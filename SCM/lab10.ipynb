## Список вагонов с `Union{Wagon, Nothing}`

```julia
struct Wagon
    cargo::Int
    next::Union{Wagon, Nothing}
end

cargo(w::Wagon) = w.cargo + cargo(w.next)
cargo(w) = 0

train = Wagon(3, Wagon(4, Wagon(1, nothing)))
cargo(train)
```

```julia
try
    Wagon(3, Wagon(4, Wagon(1, 42)))
catch e
    println(typeof(e))
end
```
