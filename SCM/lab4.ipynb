```julia
# 4.1 Анализ продажи пиццы
println("\nУпражнение 4.1: Анализ продажи пиццы")
# количеств проданных пицц
amounts = [4, 1, 5, 3, 2]
println("Количества: ", amounts)
# столбец цен
prices = [15.0, 11.5, 13.0, 12.75, 14.25]
println("Цены: ", prices)

```

```julia

# 4.2 Различные типы массивов

println("\nУпражнение 4.2: Вектор-строка и вектор-столбец")
col = [4, 1, 5, 3, 2]
println("Вектор-столбец: ", col)
row = [4 1 5 3 2]
println("Вектор-строка: ", row)

pizzas = [4 15.0;
          1 11.5;
          5 13.0;
          3 12.75;
          2 14.25]
println("Матрица пицц:")
display(pizzas)
```

```julia
# 4.3 Выполнение операций на массивах

println("\nУпражнение 4.3: Операции на массивах")
total_sold = sum(amounts)
println("Всего продано пицц: ", total_sold)

# Количество элементов
n = length(amounts)
println("Количество категорий: ", n)

# Среднее
avg = sum(amounts) / length(amounts)
println("Среднее количество: ", avg)

# Сортировка
sorted_prices = sort(prices)
println("Отсортированные цены: ", sorted_prices)

# sort! изменяет исходный массив
prices2 = [15.0, 11.5, 13.0, 12.75, 14.25]
sort!(prices2)
println("После sort!: ", prices2)

# Умножение с налогом (поэлементно)
prices_tax = prices .* 1.25
println("Цены с налогом 25%: ", prices_tax)

# Поэлементное умножение количества на цены
revenue = amounts .* prices
println("Выручка по каждой пицце: ", revenue)

println("Общая выручка: ", sum(amounts .* prices))
```

```julia
# 4.7 Отображение значений в массиве (map)


println("\nУпражнение 4.7: Функция map")
degs = [0, 15, 30, 45, 60]
rads = map(deg2rad, degs)
println("Градусы: ", degs)
println("Радианы: ", rads)

# map по диапазону
println("map(deg2rad, 0:15:90) = ", map(deg2rad, 0:15:90))

# Комбинирование map
println("map(sin, map(deg2rad, 0:15:90)) = ",
        map(sin, map(deg2rad, 0:15:90)))

# Своя функция
degsin(deg) = sin(deg2rad(deg))
println("map(degsin, 0:15:90) = ", map(degsin, 0:15:90))

# Своя реализация map
function transform(fun, xs)
    ys = []
    for x in xs
        push!(ys, fun(x))
    end
    return ys
end

println("transform(degsin, 0:15:90) = ", transform(degsin, 0:15:90))

# Функции как значения
g = sin
println("g = sin; g(1.0) = ", g(1.0))
add = +
println("add = +; add(2, 3) = ", add(2, 3))
```

```julia

# ==========================================
# 4.9 Хранение данных о пиццах в кортежах
# ==========================================

println("\nУпражнение 4.9: Кортежи")

pizza_tuple = ("hawaiian", 'S', 10.5)
println("Кортеж пиццы: ", pizza_tuple)

pza = ("hawaiian", 'S', 10.5)
println("pza[1] = ", pza[1])

for item in pza
    println("  элемент: ", item)
end

nums = (3, 4, 1)
println("sum(nums) = ", sum(nums))
println("median(nums) = ", median(nums))

# Список данных о пиццах
sales = [
    ("hawaiian", 'S', 10.5),
    ("sicilian", 'S', 12.25),
    ("hawaiian", 'L', 16.5),
    ("bbq chicken", 'L', 20.75),
    ("bbq chicken", 'M', 16.75),
]
println("Продажи: ")
for s in sales
    println("  ", s)
end

# Функции доступа
name(pizza)    = pizza[1]
portion(pizza) = pizza[2]
price(pizza)   = pizza[3]

println("map(name, sales) = ", map(name, sales))
```

```julia
# 4.10 Фильтрация пицц на основе предикатов

println("\nУпражнение 4.10: Фильтрация")
println("iseven(3) = ", iseven(3))
println("iseven(2) = ", iseven(2))
println("isodd(3)  = ", isodd(3))
println("isodd(4)  = ", isodd(4))

println("isuppercase('A') = ", isuppercase('A'))
println("isuppercase('a') = ", isuppercase('a'))
println("isspace(' ')    = ", isspace(' '))
println("isspace('X')    = ", isspace('X'))

println("filter(iseven, 1:10) = ", filter(iseven, 1:10))

issmall(pizza) = portion(pizza) == 'S'
islarge(pizza) = portion(pizza) == 'L'
isbbq(pizza)   = name(pizza) == "bbq chicken"

println("Большие пиццы: ", filter(islarge, sales))
```

```julia
# 4.10.1 Комбинирование функций более высокого порядка

println("\nУпражнение 4.10.1: Комбинирование map и filter")

large_prices = map(price, filter(islarge, sales))
println("Цены больших пицц: ", large_prices)
println("Выручка от больших пицц: ", sum(large_prices))
bbq_sales = filter(isbbq, sales)
println("Продажи bbq: ", bbq_sales)
println("Выручка от bbq: ", sum(map(price, bbq_sales)))
```

```julia
# 4.11 Отображение и редукция массива

println("\nУпражнение 4.11: mapreduce")

println("mapreduce(price, +, bbq_sales) = ", mapreduce(price, +, bbq_sales))

#mapreduce
mapcompress(f, g, xs) = reduce(g, map(f, xs))
println("mapcompress(price, +, bbq_sales) = ",
        mapcompress(price, +, bbq_sales))
println("sum(2:4)       = ", sum(2:4))
println("reduce(+, 2:4) = ", reduce(+, 2:4))
println("factorial(4)   = ", factorial(4))
println("reduce(*, 1:4) = ", reduce(*, 1:4))
```

```julia

# 4.11.1 Таблица синусов с использованием map и reduce


println("\nУпражнение 4.11.1: Синус через mapreduce")

function sinus(x)
    n = 5
    taylor(i) = (-1)^i * x^(2i+1) / factorial(2i+1)
    mapreduce(taylor, +, 0:n)
end

println("sinus(0.5) = ", sinus(0.5))
println("sin(0.5)   = ", sin(0.5))
```
