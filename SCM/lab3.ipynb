```julia
# 3.4 Многострочные функции

println("\nУпражнение 3.4: Многострочные функции")

function print_sin_table(increment, max_angle)
    angle = 0
    while angle <= max_angle
        rad = deg2rad(angle)
        println("sin($angle °) = ", sin(rad))
        angle += increment
    end
end

print_sin_table(30, 90)
```

```julia
function my_sin(x)
    n = 5
    total = 0
    for i in 0:n
        total += (-1)^i * x^(2i+1) / factorial(2i+1)
    end
    return total
end
```

```julia
println("\nУпражнение 3.5: Факториал")

function my_factorial(n)
    result = 1
    while n >= 1
        result *= n
        n -= 1
    end
    return result
end

println("Факториал 5 (цикл): ", my_factorial(5))
println("Факториал 5 (встроенный): ", factorial(5))
```

```julia
println("\nУпражнение 3.6: Рекурсивный факториал")

function rec_factorial(n)
    if n <= 1
        return 1
    else
        return n * rec_factorial(n - 1)
    end
end

println("Рекурсивный факториал 5: ", rec_factorial(5))
```
