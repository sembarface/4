```julia
println("\nУпражнение 6.1: Разбор римских цифр")

# Словарь: римская буква => десятичное значение
roman_numerals = Dict(
    'I' => 1,
    'V' => 5,
    'X' => 10,
    'L' => 50,
    'C' => 100,
    'D' => 500,
    'M' => 1000,
)

println("Словарь римских цифр: ", roman_numerals)
# Функция конвертации
function parse_roman(s)
    s = reverse(uppercase(s))                     # переворачиваем, в верхний регистр
    vals = [roman_numerals[ch] for ch in s]       # массив значений
    result = 0
    for (i, val) in enumerate(vals)
        if i > 1 && val < vals[i - 1]
            result -= val
        else
            result += val
        end
    end
    return result
end
# Проверка
for r in ["II", "IV", "VI", "IX", "XI", "XIV", "MCMXCIV", "MMXXIV"]
    println("parse_roman(\"$r\") = ", parse_roman(r))
end
```

```julia
println("\nУпражнение 6.5: Как работает вычитание")

# XVI = 16, но XIV = 14 (I стоит перед V)
println("XVI = ", parse_roman("XVI"))
println("XIV = ", parse_roman("XIV"))

# Перевёрнутая обработка: vals и сравнение с предыдущим
vals = [10, 1, 5]   # XIV в перевёрнутом виде [V, I, X] -> [5, 1, 10]?
println("Разберём XIV:")
s = reverse(uppercase("XIV"))   # "VIX"
println("  Перевёрнутая строка: ", s)
vals = [roman_numerals[ch] for ch in s]
println("  Значения: ", vals)
result = 0
for (i, val) in enumerate(vals)
    if i > 1 && val < vals[i - 1]
        result -= val
        println("  i=$i: val=$val < предыдущего, вычитаем -> $result")
    else
        result += val
        println("  i=$i: val=$val >= предыдущего, прибавляем -> $result")
    end
end
println("  Итог: ", result)
```

```julia
println("\nУпражнение 6.6: Создание и использование словарей")

# Способ 1: с парами
d1 = Dict("two" => 2, "four" => 4)
println("d1 = ", d1)

# Способ 2: из массива пар
pairs = ["two" => 2, "four" => 4]
d2 = Dict(pairs)
println("d2 = ", d2)

# Способ 3: из массива кортежей
tuples = [("two", 2), ("four", 4)]
d3 = Dict(tuples)
println("d3 = ", d3)
# Из данных о пиццах
pizzas = [
    ("mexicana",    13.0),
    ("hawaiian",    16.5),
    ("bbq chicken", 20.75),
    ("sicilian",    12.25),
]
pizza_dict = Dict(pizzas)
println("pizza_dict = ", pizza_dict)
println("pizza_dict[\"mexicana\"] = ", pizza_dict["mexicana"])

# Пустой словарь
d = Dict{String, Float64}()
println("Пустой словарь: ", d)
# Добавление в пустой словарь
d["hawaiian"] = 16.5
d["mexicana"] = 13.0
println("После добавления: ", d)

# Неверный тип ключа вызовет ошибку (закомментировано):
# d[5] = "five"

# Создание из двух массивов через zip
words = ["one", "two"]
nums  = [1, 2]
println("collect(zip(words, nums)) = ", collect(zip(words, nums)))
println("Dict(zip(words, nums))    = ", Dict(zip(words, nums)))
```

```julia
println("\nУпражнение 6.7: Почему словарь быстрее массива")

# Реализация поиска в массиве пар
function lookup(key, table)
    for (k, v) in table
        if key == k
            return v
        end
    end
    throw(KeyError(key))
end
# Сверочная таблица в виде массива пар
numerals = [
    'I' => 1,
    'X' => 10,
    'C' => 100,
    'V' => 5,
    'L' => 50,
    'D' => 500,
    'M' => 1000,
]

println("lookup('X', numerals) = ", lookup('X', numerals))
println("lookup('D', numerals) = ", lookup('D', numerals))

# Несуществующий ключ
try
    println(lookup('S', numerals))
catch e
    println("Поймана ошибка: ", e)
end

# Массив отсортированных ключей + массив значений
keys_arr = ['C', 'D', 'I', 'L', 'M', 'V', 'X']
vals_arr = [100, 500, 1, 50, 1000, 5, 10]

i = searchsortedfirst(keys_arr, 'I')
println("Индекс 'I' в отсортированном массиве: ", i)
println("Значение: ", vals_arr[i])
```

```julia
println("\nУпражнение 6.8: Именованные кортежи (NamedTuple)")

# Обычный кортеж
pizza_tuple = ("hawaiian", 'S', 10.5)
println("Обычный кортеж: ", pizza_tuple)

# Именованный кортеж
pizza_named = (name = "hawaiian", size = 'S', price = 10.5)
println("Именованный кортеж: ", pizza_named)

# Доступ по имени
println("pizza_named[:name]   = ", pizza_named[:name])
println("pizza_named[:price]  = ", pizza_named[:price])
println("pizza_named.name     = ", pizza_named.name)
println("pizza_named.size     = ", pizza_named.size)

# Конвертация между Symbol и String
s = "price"
t = :name
println("Symbol(s) = ", Symbol(s))
println("string(t) = ", string(t))
```

```julia
println("\nУпражнение 6.8.2: Разбор римских цифр через NamedTuple")

# Именованный кортеж в качестве сверочной таблицы
roman_numerals_nt = (
    I = 1,
    X = 10,
    C = 100,
    V = 5,
    L = 50,
    D = 500,
    M = 1000,
)

function parse_roman_nt(s)
    s = reverse(uppercase(s))
    vals = [roman_numerals_nt[Symbol(ch)] for ch in s]
    result = 0
    for (i, val) in enumerate(vals)
        if i > 1 && val < vals[i - 1]
            result -= val
        else
            result += val
        end
    end
    return result
end

# Проверка
for r in ["II", "IV", "VI", "IX", "XI", "XIV", "MCMXCIV"]
    println("parse_roman_nt(\"$r\") = ", parse_roman_nt(r))
end
```

```julia
println("\nСравнение производительности")

# Готовим тестовые строки
test_romans = ["MCMXCIV", "MMXXIV", "XIV", "IX", "XLII", "CDXLIV"]

# Dict
t1 = @elapsed for _ in 1:1000, r in test_romans
    parse_roman(r)
end
println("Dict:        $t1 сек")

# NamedTuple
t2 = @elapsed for _ in 1:1000, r in test_romans
    parse_roman_nt(r)
end
println("NamedTuple:  $t2 сек")
```
