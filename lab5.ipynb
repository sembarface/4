```julia
println("\nУпражнение 5.1.4: lpad и rpad")

println("|", lpad("ABC", 6, '-'), "|")
println("|", rpad("ABC", 6, '-'), "|")
println("|", lpad("ABC", 6), "|")
println("|", rpad("ABC", 6), "|")

# Проверка ширины столбцов
println("length(\"thai chicken\") = ", length("thai chicken"))
println("length(\"size\") = ", length("size"))
println("max(length(\"16.75\"), length(\"price\")) = ",
        max(length("16.75"), length("price")))

# Простая выровненная таблица
function simple_pizzatable(pizzas)
    pname  = rpad("name", 12)
    psize  = rpad("size", 4)
    pprice = rpad("price", 5)
    printstyled(pname, " ", psize, " ", pprice, "\n", color = :cyan)
    println()
    for pz in pizzas
        pname  = rpad(name(pz), 12)
        psize  = rpad(portion(pz), 4)
        pprice = lpad(price(pz), 5)
        println(pname, " ", psize, " ", pprice)
    end
end

simple_pizzatable(sales)
```

```julia
println("\nУпражнение 5.1.5: Разделительные линии")

println('|')
println("Символ: ", '|')
println("Код символа: ", Int('|'))

# Таблица с разделителем '|'
function pizzatable_with_lines(pizzas)
    println(rpad("name", 12), " | ", rpad("size", 4), " | ", rpad("price", 5))
    println("-"^25)
    for pz in pizzas
        println(rpad(name(pz), 12), " | ",
                rpad(portion(pz), 4), " | ",
                lpad(price(pz), 5))
    end
end

pizzatable_with_lines(sales)
```

```julia
println("\nУпражнение 5.2: Тригонометрическая таблица")

# Форматирование чисел
n = length(" -0.966")   # максимальная ширина

function format_number(x)
    x = round(x, digits = 3)
    if x < 0
        rpad(x, n, '0')
    else
        rpad(x, n - 1, '0')
    end
end

println("format_number(3.1)  = ", format_number(3.1))
println("format_number(-3.1) = ", format_number(-3.1))
function print_trigtable(inc, maxangle)
    println(rpad("θ", 3), " | ", lpad("cos", 6), " | ", lpad("sin", 6))
    println("-"^23)
    angle = 0
    while angle <= maxangle
        rad  = deg2rad(angle)
        cosx = format_number(cos(rad))
        sinx = format_number(sin(rad))
        println(lpad(angle, 3), " | ", lpad(cosx, 6), " | ", lpad(sinx, 6))
        angle += inc
    end
end

print_trigtable(15, 90)
```

```julia
println("\nУпражнение 5.3: CSV-файлы")

# Создаём CSV-файл для последующих упражнений
function store_pizzatable(io, pizzas)
    println(io, "name,size,price")
    for pz in pizzas
        println(io, name(pz), ",", portion(pz), ",", price(pz))
    end
end

# Записываем в файл
io = open("pizza-sales.csv", "w")
store_pizzatable(io, sales)
close(io)
println("Файл pizza-sales.csv создан.")
# Смотрим содержимое
println("\nСодержимое файла pizza-sales.csv:")
for line in readlines("pizza-sales.csv")
    println("  ", line)
end
```

```julia
println("\nУпражнение 5.3.2: Чтение из файла")

# Чтение с клавиатуры (не выполнять в Jupyter, оставлено для примера)
# s = readline(stdin)
# println("Вы ввели: ", s)

# Чтение из файла построчно
io = open("pizza-sales.csv")
line = readline(io)
println("Первая строка (заголовок): ", line)
line = readline(io)
println("Вторая строка: ", line)
close(io)

# Разбор строки с помощью split
pizza = split("hawaiian,S,10.5", ',')
println("split: ", pizza)
println("pizza[1] = ", pizza[1])
println("pizza[3] = ", pizza[3])

# Превращение строки в число
p = pizza[3]
println("typeof(p) = ", typeof(p))
println("parse(Float64, p) = ", parse(Float64, p))

# Демонстрация parse
println("parse(Int, \"42\")      = ", parse(Int, "42"))
println("parse(Float64, \"42\")  = ", parse(Float64, "42"))
println("parse(Bool, \"true\")   = ", parse(Bool, "true"))
println("parse(Bool, \"1\")      = ", parse(Bool, "1"))
# Полная функция загрузки
function load_pizzatable(io)
    pizzas = []
    readline(io)   # пропустить заголовок
    while !eof(io)
        pz = split(readline(io), ',')
        pr = parse(Float64, pz[3])
        sz = pz[2][1]           # берём первый символ как Char
        push!(pizzas, (pz[1], sz, pr))
    end
    return pizzas
end

io = open("pizza-sales.csv")
loaded = load_pizzatable(io)
close(io)
println("Загружено пицц: ", length(loaded))
for pz in loaded
    println("  ", pz)
end
```

```julia
println("\nУпражнение 5.4: Интерактивное приложение (таблица умножения)")

# Автоматический вариант без stdin — для Jupyter удобнее
function practice_auto(n)
    correct = 0
    for i in 1:n
        x = rand(2:9)
        y = rand(2:9)
        answer = x * y   # имитируем правильный ответ
        print(x, " * ", y, " = ", answer)
        if answer == x * y
            correct += 1
        else
            printstyled("  Wrong, it is ", x * y, "\n", color = :red)
        end
        println()
    end
    println("Correct: ", correct, " of ", n)
end
practice_auto(4)
```
